```polyglot-notebook

using System;

// Интерфейс для всех животных
public interface IAnimal
{
    void MakeSound();
    void Eat();
    string GetName();
    void ShowInfo();
}

// Абстрактный базовый класс Animal
public abstract class Animal : IAnimal
{
    // Свойства с геттерами и сеттерами
    public string Name { get; set; }
    public int Age { get; set; }
    public double Weight { get; set; }

    // Конструктор
    public Animal(string name, int age, double weight)
    {
        Name = name;
        Age = age;
        Weight = weight;
    }

    // Абстрактный метод
    public abstract void MakeSound();

    // Виртуальный метод
    public virtual void Eat()
    {
        Console.WriteLine($"{Name} ест.");
    }

    // Метод получения имени
    public string GetName()
    {
        return Name;
    }

    // Метод вывода информации
    public virtual void ShowInfo()
    {
        Console.WriteLine($"Имя: {Name}");
        Console.WriteLine($"Возраст: {Age} лет");
        Console.WriteLine($"Вес: {Weight} кг");
    }
}

// Производный класс Dog
public class Dog : Animal
{
    public string Breed { get; set; }

    public Dog(string name, int age, double weight, string breed)
        : base(name, age, weight)
    {
        Breed = breed;
    }

    public override void MakeSound()
    {
        Console.WriteLine($"{Name} говорит: Гав-гав!");
    }

    public override void Eat()
    {
        Console.WriteLine($"{Name} ест собачий корм.");
    }

    public override void ShowInfo()
    {
        base.ShowInfo();
        Console.WriteLine($"Порода: {Breed}");
    }

    public void GuardHouse()
    {
        Console.WriteLine($"{Name} охраняет дом.");
    }
}

// Производный класс Cat
public class Cat : Animal
{
    public string Color { get; set; }

    public Cat(string name, int age, double weight, string color)
        : base(name, age, weight)
    {
        Color = color;
    }

    public override void MakeSound()
    {
        Console.WriteLine($"{Name} говорит: Мяу!");
    }

    public override void Eat()
    {
        Console.WriteLine($"{Name} ест рыбу.");
    }

    public override void ShowInfo()
    {
        base.ShowInfo();
        Console.WriteLine($"Цвет: {Color}");
    }

    public void Climb()
    {
        Console.WriteLine($"{Name} лазает по деревьям.");
    }
}

// Производный класс Bird
public class Bird : Animal
{
    public double WingSpan { get; set; }

    public Bird(string name, int age, double weight, double wingSpan)
        : base(name, age, weight)
    {
        WingSpan = wingSpan;
    }

    public override void MakeSound()
    {
        Console.WriteLine($"{Name} поёт: Чирик-чирик!");
    }

    public override void Eat()
    {
        Console.WriteLine($"{Name} клюёт зерно.");
    }

    public override void ShowInfo()
    {
        base.ShowInfo();
        Console.WriteLine($"Размах крыльев: {WingSpan} м");
    }

    public void Fly()
    {
        Console.WriteLine($"{Name} летает.");
    }
}

// Создание объектов
IAnimal myDog = new Dog("Шарик", 3, 12.5, "Овчарка");
IAnimal myCat = new Cat("Мурка", 2, 4.2, "Белая");
IAnimal myBird = new Bird("Кеша", 1, 0.5, 0.3);

// Массив животных через интерфейс
IAnimal[] animals = { myDog, myCat, myBird };

// Вывод информации обо всех животных
foreach (IAnimal animal in animals)
{
    Console.WriteLine("\n============================");
    animal.ShowInfo();
    animal.MakeSound();
    animal.Eat();
}

// Дополнительные действия
Console.WriteLine("\nДополнительные действия:");
Console.WriteLine("============================");

((Dog)myDog).GuardHouse();
((Cat)myCat).Climb();
((Bird)myBird).Fly();

Console.WriteLine("\nПрограмма завершена!");
```
