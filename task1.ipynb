```polyglot-notebook

using System;

// Интерфейс для всех животных
public interface IAnimal
{
    void MakeSound();
    void Eat();
    string GetName();
    void ShowInfo();
}

// Интерфейс для домашних и прирученных животных
public interface IPet : IAnimal
{
    void Play();
}

// Абстрактный базовый класс
public abstract class Animal : IAnimal
{
    public string Name { get; set; }
    public int Age { get; set; }
    public double Weight { get; set; }
    public string Habitat { get; set; }

    public Animal(string name, int age, double weight, string habitat)
    {
        Name = name;
        Age = age;
        Weight = weight;
        Habitat = habitat;
    }

    // Абстрактный метод: каждый вид издаёт свой звук
    public abstract void MakeSound();

    // Виртуальный метод питания
    public virtual void Eat()
    {
        Console.WriteLine($"{Name} принимает пищу.");
    }

    // Явная реализация интерфейса
    string IAnimal.GetName()
    {
        return Name;
    }

    // Вывод информации
    public void ShowInfo()
    {
        Console.WriteLine($"Имя: {Name}");
        Console.WriteLine($"Возраст: {Age} лет");
        Console.WriteLine($"Вес: {Weight} кг");
        Console.WriteLine($"Среда обитания: {Habitat}");
    }
}

// Лев
public class Lion : Animal
{
    public int ManeLength { get; set; }

    public Lion(string name, int age, double weight, string habitat, int maneLength)
        : base(name, age, weight, habitat)
    {
        ManeLength = maneLength;
    }

    public override void MakeSound()
    {
        Console.WriteLine($"{Name} громко рычит: Р-Р-Р!");
    }

    public override void Eat()
    {
        Console.WriteLine($"{Name} охотится и ест мясо.");
    }

    public void ProtectTerritory()
    {
        Console.WriteLine($"{Name} защищает свою территорию!");
    }

    public void ShowMane()
    {
        Console.WriteLine($"Длина гривы льва: {ManeLength} см.");
    }
}

// Пингвин
public class Penguin : Animal, IPet
{
    public double SwimSpeed { get; set; }

    public Penguin(string name, int age, double weight, string habitat, double swimSpeed)
        : base(name, age, weight, habitat)
    {
        SwimSpeed = swimSpeed;
    }

    public override void MakeSound()
    {
        Console.WriteLine($"{Name} издаёт характерный крик: Кря-кря!");
    }

    public override void Eat()
    {
        Console.WriteLine($"{Name} ловит и ест рыбу.");
    }

    public void Swim()
    {
        Console.WriteLine($"{Name} плавает со скоростью {SwimSpeed} км/ч.");
    }

    public void Play()
    {
        Console.WriteLine($"{Name} играет с мячом!");
    }
}

// Черепаха
public class Turtle : Animal
{
    public int ShellHardness { get; set; }

    public Turtle(string name, int age, double weight, string habitat, int shellHardness)
        : base(name, age, weight, habitat)
    {
        ShellHardness = shellHardness;
    }

    public override void MakeSound()
    {
        Console.WriteLine($"{Name} тихо шипит.");
    }

    public override void Eat()
    {
        Console.WriteLine($"{Name} медленно ест растения.");
    }

    public void HideInShell()
    {
        Console.WriteLine($"{Name} прячется в свой панцирь!");
    }

    public void ShowShell()
    {
        Console.WriteLine($"Прочность панциря: {ShellHardness}/100.");
    }
}

// Создание объектов
Lion lion = new Lion("Симба", 5, 190, "Саванна", 25);
Penguin penguin = new Penguin("Пин", 3, 18, "Антарктида", 12);
Turtle turtle = new Turtle("Тортилла", 40, 80, "Тропический лес", 95);

// Работа через интерфейс IAnimal
Console.WriteLine("========== ЖИВОТНЫЕ ==========");

IAnimal animal1 = lion;
IAnimal animal2 = penguin;
IAnimal animal3 = turtle;

animal1.ShowInfo();
animal1.MakeSound();
animal1.Eat();

Console.WriteLine();

animal2.ShowInfo();
animal2.MakeSound();
animal2.Eat();

Console.WriteLine();

animal3.ShowInfo();
animal3.MakeSound();
animal3.Eat();

// Вызов явной реализации интерфейса
Console.WriteLine("\n========== ИМЕНА ЧЕРЕЗ ИНТЕРФЕЙС ==========");
Console.WriteLine(animal1.GetName());
Console.WriteLine(animal2.GetName());
Console.WriteLine(animal3.GetName());

// Уникальные действия животных
Console.WriteLine("\n========== ОСОБЕННОСТИ ЖИВОТНЫХ ==========");

lion.ProtectTerritory();
lion.ShowMane();

penguin.Swim();
penguin.Play();

turtle.HideInShell();
turtle.ShowShell();
```
