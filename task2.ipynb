```polyglot-notebook

using System;

// Интерфейс для всех животных
public interface IAnimal
{
    void MakeSound();
    void Eat();
    string GetName();
    void ShowInfo();
}

// Абстрактный базовый класс
public abstract class Animal : IAnimal
{
    // Свойства с геттерами и сеттерами
    public string Name { get; set; }
    public int Age { get; set; }
    public double Weight { get; set; }
    public string Habitat { get; set; }

    // Конструктор
    public Animal(string name, int age, double weight, string habitat)
    {
        Name = name;
        Age = age;
        Weight = weight;
        Habitat = habitat;
    }

    // Абстрактный метод
    public abstract void MakeSound();

    // Виртуальный метод
    public virtual void Eat()
    {
        Console.WriteLine($"{Name} принимает пищу.");
    }

    // Реализация метода интерфейса
    public string GetName()
    {
        return Name;
    }

    // Вывод информации
    public void ShowInfo()
    {
        Console.WriteLine($"Имя: {Name}");
        Console.WriteLine($"Возраст: {Age} лет");
        Console.WriteLine($"Вес: {Weight} кг");
        Console.WriteLine($"Среда обитания: {Habitat}");
    }
}

// Класс Лев
public class Lion : Animal
{
    public int ManeLength { get; set; }

    public Lion(string name, int age, double weight,
                string habitat, int maneLength)
        : base(name, age, weight, habitat)
    {
        ManeLength = maneLength;
    }

    public override void MakeSound()
    {
        Console.WriteLine($"{Name} громко рычит: Р-Р-Р!");
    }

    public override void Eat()
    {
        Console.WriteLine($"{Name} охотится и ест мясо.");
    }

    public void ProtectTerritory()
    {
        Console.WriteLine($"{Name} защищает свою территорию!");
    }
}

// Класс Пингвин
public class Penguin : Animal
{
    public double SwimSpeed { get; set; }

    public Penguin(string name, int age, double weight,
                   string habitat, double swimSpeed)
        : base(name, age, weight, habitat)
    {
        SwimSpeed = swimSpeed;
    }

    public override void MakeSound()
    {
        Console.WriteLine($"{Name} издаёт характерный крик!");
    }

    public override void Eat()
    {
        Console.WriteLine($"{Name} ловит и ест рыбу.");
    }

    public void Swim()
    {
        Console.WriteLine($"{Name} плавает со скоростью {SwimSpeed} км/ч.");
    }
}

// Класс Черепаха
public class Turtle : Animal
{
    public int ShellHardness { get; set; }

    public Turtle(string name, int age, double weight,
                  string habitat, int shellHardness)
        : base(name, age, weight, habitat)
    {
        ShellHardness = shellHardness;
    }

    public override void MakeSound()
    {
        Console.WriteLine($"{Name} тихо шипит.");
    }

    public override void Eat()
    {
        Console.WriteLine($"{Name} медленно ест растения.");
    }

    public void HideInShell()
    {
        Console.WriteLine($"{Name} прячется в свой панцирь!");
    }
}

// Сервис для работы с животными
public class AnimalService
{
    private readonly IAnimal _animal;

    // Внедрение зависимости через конструктор
    public AnimalService(IAnimal animal)
    {
        _animal = animal;
    }

    // Использование переданного животного
    public void UseAnimal()
    {
        Console.WriteLine("\n========== ИНФОРМАЦИЯ О ЖИВОТНОМ ==========");

        _animal.ShowInfo();

        Console.WriteLine($"Имя через интерфейс: {_animal.GetName()}");

        _animal.MakeSound();
        _animal.Eat();
    }
}

// Создание объектов
Lion lion = new Lion("Симба", 5, 190, "Саванна", 25);

Penguin penguin = new Penguin("Пин", 3, 18, "Антарктида", 12);

Turtle turtle = new Turtle("Тортилла", 40, 80, "Тропический лес", 95);

// Создание сервисов с внедрением зависимостей
AnimalService lionService = new AnimalService(lion);

AnimalService penguinService = new AnimalService(penguin);

AnimalService turtleService = new AnimalService(turtle);

// Использование сервисов
lionService.UseAnimal();

penguinService.UseAnimal();

turtleService.UseAnimal();

// Вызов уникальных методов животных
Console.WriteLine("\n========== ОСОБЕННОСТИ ЖИВОТНЫХ ==========");

lion.ProtectTerritory();

penguin.Swim();

turtle.HideInShell();
```
